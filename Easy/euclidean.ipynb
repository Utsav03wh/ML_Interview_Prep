X ∈ ℝ^(N, D) — N vectors of dimension D

Y ∈ ℝ^(M, D) — M vectors of dimension D

D[i, j] = ||X[i] - Y[j]||₂ = sqrt( Σ_d (X[i,d] - Y[j,d])² )

||x - y||² = ||x||² - 2⟨x, y⟩ + ||y||²



```python
import torch 
import torch.nn.functional as F 

def pairwise_euclidean_dist(X,Y,squared):
    """
    Compute pairwise Euclidean distances between rows of X and Y.
    Math:
        ||x - y||² = ||x||² - 2⟨x, y⟩ + ||y||²
    This avoids loops entirely and is the standard way to do it.
    Args:
        X: [N, D]
        Y: [M, D]
        squared: If True, return squared distances (avoids sqrt)
    Returns:
        D: [N, M]  where D[i, j] = ||X[i] - Y[j]||₂ (or squared)
    """

    x_norm_sq = torch.square(X).sum(dim=1).unsqueeze(1) # [N,1]
    y_norm_sq = torch.square(Y).sum(dim=-1).unsqueeze(1) # [M,1]

    cross_term = X @ Y.transpose(0,1) # [N,M]

    dist = x_norm_sq + y_norm_sq -2*cross_term # [N,M]

    return dist.sqrt()  



    
```

# torch.scatter 
torch.scatter(input, dim, index, src) -> Tensor
• input (Tensor): The base destination tensor that values will be written into.

• dim (int): The axis along which the indexing happens.

• index (LongTensor): The tensor containing the indices of elements to scatter. Its dimensions must match the structure of src (or be broadcastable).

• src (Tensor or scalar): The source tensor containing the values to write, or a scalar value to fill into the designated positions

```python
def one_hot_scatter(targets,num_classes):

    """
    targets [N]
    num_classes = C
    returns [N,C
    
    """

    N = targets.shape[0]

    one_hot = torch.zeros(N,num_classes) # N,C

    one_hot.scatter_(dim=1,index=targets.unsqueeze(1),src=1.0)


    return one_hot
```
